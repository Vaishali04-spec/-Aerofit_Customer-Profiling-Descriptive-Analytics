```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#read dataset
df = pd.read_csv('/content/aerofit_treadmill.csv')
df.info()
```

```python
data = df.dropna()
data.info()
```

```python
y_count = data['MaritalStatus'].value_counts()
plt.bar(y_count.index, y_count)
plt.title('Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Count')
```

```python
data_age = data[data['MaritalStatus'] == 'Single']
```

```python
plt.figure(figsize=(5,4))
sns.histplot(data_age['Gender'], bins= 10)
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Frequency')
```

Above plot depicts that bachelor males are more active than female.

```python
# age distribution of Singele Male
active_age = data_age[data_age['Gender']== 'Male']
active_age1 = active_age['Age'].value_counts()
plt.figure(figsize=(8,4))
sns.barplot(data = active_age1, color = 'red')
```

```python
#Age distribution of single female
active_age = data_age[data_age['Gender']== 'Female']
active_age1 = active_age['Age'].value_counts()
plt.figure(figsize=(8,4))
sns.barplot(data = active_age1, color = 'blue')
```

```python
# percentage of female
perc_female = data[data['Gender']== 'Female'].shape[0]/data.shape[0]*100.
print(round(perc_female,2))
```

```python
data_agg = data.groupby(['Product'])[['Age']].aggregate({'Age': 'mean'})
data_agg
```

From the above table, it suggests that folks in lates 20's prefer or can afford the more advanced or expensive treadmill models.Also indicate more health-conscious or experienced users moving toward advanced products.

```python
# percent of customers have purchased KP281, KP481, or KP781
num_prod = pd.crosstab(index=data['Product'], columns = 'counts')
perc = num_prod.apply(lambda x:x/x.sum()*100).round(2)
perc
```

```python
#Visual representation percent of customers have purchased KP281, KP481, or KP781(Marginal probability)
fig, ax = plt.subplots()
ax.pie(perc['counts'], labels = perc.index, autopct = '%1.1f%%')
```

```python
num_prod = pd.crosstab(index=data['Product'], columns = data['Gender'], margins = True)
num_prod
```

```python
# probability that a customer is female given they purchased the KP281 model?
prob_cond = data[(data['Gender']== 'Female') & (data['Product']== 'KP281')].shape[0]/data[data['Product']== 'KP281'].shape[0]*100
prob_cond
```

```python
#What is the probability of a male customer buying a KP781 treadmill?
prob_male = data[(data['Gender']== 'Male')& (data['Product']== 'KP781')].shape[0]/data[data['Product']== 'KP781'].shape[0]*100
prob_male
```
